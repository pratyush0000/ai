```python
import numpy as np
import pandas as pd
```

```python
# Initial population
population = {
    "x1": [6, 5, 4, 1, 3, 5, 3, 2],
    "x2": [8, 7, 1, 2, 6, 6, 0, 1],
    "x3": [2, 3, 9, 2, 1, 2, 8, 5],
    "x4": [4, 1, 8, 5, 2, 0, 9, 4],
}

# Fitness calculation
def calculate_fitness(chromosome):
    return (chromosome[0] + chromosome[1]) - (chromosome[2] + chromosome[3]) + \
           (chromosome[4] + chromosome[5]) - (chromosome[6] + chromosome[7])

# Evaluate fitness for all individuals
fitness_scores = {key: calculate_fitness(value) for key, value in population.items()}

# Rank individuals by fitness
fitness_df = pd.DataFrame(list(fitness_scores.items()), columns=["Individual", "Fitness"])
fitness_df["Chromosome"] = list(population.values())
fitness_df = fitness_df.sort_values(by="Fitness", ascending=False).reset_index(drop=True)

print("Fitness Ranking:")
print(fitness_df)

# Select the top individuals for crossover
fittest = fitness_df.loc[0, "Chromosome"]
second_fittest = fitness_df.loc[1, "Chromosome"]
third_fittest = fitness_df.loc[2, "Chromosome"]

# One-point crossover
mid_point = len(fittest) // 2
child1 = fittest[:mid_point] + second_fittest[mid_point:]
child2 = second_fittest[:mid_point] + fittest[mid_point:]

# Two-point crossover
b, f = 1, 5  # Adjust 1-based to 0-based indexing
child3 = second_fittest[:b] + third_fittest[b:f] + second_fittest[f:]
child4 = third_fittest[:b] + second_fittest[b:f] + third_fittest[f:]

print("\nOne-Point Crossover:")
print("Child1:", child1)
print("Child2:", child2)

print("\nTwo-Point Crossover:")
print("Child3:", child3)
print("Child4:", child4)
```
